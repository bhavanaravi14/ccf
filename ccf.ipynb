```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("creditcard.csv")
```

```python
df.info()
```

```python
if df.isnull().sum().sum() > 0:
   df.fillna(method= "ffill",inplace="True")
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x="Class" , data=df, palette="Set2")
plt.title("Fraud Vs Non-Fraud Transactions")
plt.xlabel("Transaction Class(0=Non-Fraud, 1=Fraud)")
plt.ylabel("count")
plt.grid(axis="y", linestyle="--", alpha = 0.7)
plt.tight_layout()
plt.show()
           
```

```python
plt.figure(figsize=(20,12))
corr = df.corr()
```

```python
sorted_corr = corr["Class"].abs().sort_values(ascending=False)

sns.heatmap(corr[sorted_corr.index][:20] , cmap='coolwarm', annot=False, linewidths=0.5)
plt.title("Top correlation heatmap with Class")
plt.tight_layout()
plt.show()
```

```python
data=pd.read_csv("creditcard.csv")
```

```python
data.drop(["Time"], axis=1,inplace=True)
```

```python
data["LogAmount"] = np.log1p(data["Amount"])
```

```python
data.drop(["Amount"],axis=1,inplace=True)
```

```python
plt.figure(figsize=(6,4))
sns.histplot(data["LogAmount"],bins=90,kde=True)
plt.title("Distribution of Log Transformed Amount")
plt.show()
```
